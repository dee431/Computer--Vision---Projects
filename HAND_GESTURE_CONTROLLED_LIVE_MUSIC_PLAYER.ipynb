```python
import json
from IPython.display import HTML, display
import yt_dlp

# 20 Tracks Catalog
tracks = {
    "1. Brown Rang": "https://www.youtube.com/watch?v=PqFMFV91a20",
    "2. Desi Kalakaar": "https://www.youtube.com/watch?v=KhnVcAC5bIM",
    "3. Blue Eyes": "https://www.youtube.com/watch?v=NbyHNASFi6U",
    "4. Love Dose": "https://www.youtube.com/watch?v=Kc1Jq1Yd4v4",
    "5. Dope Shope": "https://www.youtube.com/watch?v=nR4S_4_0Tno",
    "6. Lungi Dance": "https://www.youtube.com/watch?v=69CEiHU8kNE",
    "7. High Heels": "https://www.youtube.com/watch?v=52E_1L9i1m8",
    "8. Angreji Beat": "https://www.youtube.com/watch?v=V3f9R1qD3i8",
    "9. Chaar Botal Vodka": "https://www.youtube.com/watch?v=x2I--4x7sEE",
    "10. Sunny Sunny": "https://www.youtube.com/watch?v=C3171g3XG7U",
    "11. Dheere Dheere": "https://www.youtube.com/watch?v=nCD2hj6zJEc",
    "12. Yaar Naa Miley": "https://www.youtube.com/watch?v=48010C4nUf0",
    "13. Dil Chori": "https://www.youtube.com/watch?v=xWi8nDUjH50",
    "14. Chhote Chhote Peg": "https://www.youtube.com/watch?v=13I6p0qS3dI",
    "15. Makhna": "https://www.youtube.com/watch?v=HqUeS3A1K_I",
    "16. One Bottle Down": "https://www.youtube.com/watch?v=83pEChYc7E4",
    "17. Kalaastar": "https://www.youtube.com/watch?v=K1C4-a1_qfI",
    "18. Saiyaan Ji": "https://www.youtube.com/watch?v=1-3Q8p1s6L0",
    "19. Millionaire": "https://www.youtube.com/watch?v=0_uP0yJ1Lg4",
    "20. Payal": "https://www.youtube.com/watch?v=p4S9A81A9sM"
}

print("⏳ Resolving and extracting live audio stream links for all 20 tracks...")
print("ℹ️ Note: If an official video is unavailable, we search and extract an alternate copy automatically.")

audio_streams = {}
ydl_opts = {'format': 'bestaudio/best', 'quiet': True, 'no_warnings': True}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    for title, url in tracks.items():
        try:
            # Try extracting directly first
            info = ydl.extract_info(url, download=False)
            audio_streams[title] = info['url']
            print(f"✅ Successfully resolved: {title}")
        except Exception:
            # Fallback search if link is unavailable or dead
            try:
                search_query = f"ytsearch1:Yo Yo Honey Singh {title[3:]} audio"
                info = ydl.extract_info(search_query, download=False)
                if 'entries' in info and len(info['entries']) > 0:
                    audio_streams[title] = info['entries'][0]['url']
                    print(f"🔄 Restored using alternative search: {title}")
                else:
                    raise Exception("No search results")
            except Exception as e:
                print(f"❌ Failed to resolve {title}: {e}")

tracks_json = json.dumps(audio_streams)

html_code = f"""
<div style="font-family: Arial, sans-serif; background: #121212; color: white; padding: 20px; border-radius: 12px; max-width: 650px; margin: auto;">
    <h2 style="text-align: center; color: #1DB954; margin-top: 0;">🎵 Gesture-Controlled Live Music Player</h2>

    <div style="margin-bottom: 15px; text-align: center;">
        <label for="songSelect" style="font-weight: bold; margin-right: 10px;">Select Song:</label>
        <select id="songSelect" style="padding: 8px 12px; border-radius: 6px; background: #282828; color: white; border: 1px solid #404040; font-size: 14px; width: 60%;">
        </select>
    </div>

    <div style="text-align: center; margin-bottom: 15px;">
        <audio id="audioPlayer" controls style="width: 85%;"></audio>
    </div>

    <div style="display: flex; justify-content: center; align-items: center; position: relative;">
        <video id="webcam" style="display: none;" autoplay playsinline></video>
        <canvas id="canvas" width="480" height="360" style="border: 2px solid #1DB954; border-radius: 8px; background: #000;"></canvas>
    </div>

    <div style="margin-top: 15px; text-align: center;">
        <h3 id="volDisplay" style="color: #00E676; margin: 5px 0;">Volume: 100%</h3>
        <p style="font-size: 13px; color: #b3b3b3;">Pinch thumb & index finger to lower volume | Spread them apart to increase volume.</p>
    </div>
</div>

<!-- Load MediaPipe Libraries -->
<script src="https://cdn.jsdelivr.net/npm/@mediapipe/camera_utils/camera_utils.js" crossorigin="anonymous"></script>
<script src="https://cdn.jsdelivr.net/npm/@mediapipe/hands/hands.js" crossorigin="anonymous"></script>

<script>
const songData = {tracks_json};
const songSelect = document.getElementById('songSelect');
const audioPlayer = document.getElementById('audioPlayer');
const volDisplay = document.getElementById('volDisplay');
const videoElement = document.getElementById('webcam');
const canvasElement = document.getElementById('canvas');
const canvasCtx = canvasElement.getContext('2d');

// Populate Dropdown Menu
for (let title in songData) {{
    let option = document.createElement('option');
    option.value = songData[title];
    option.text = title;
    songSelect.appendChild(option);
}}

// Set initial track
if (songSelect.options.length > 0) {{
    audioPlayer.src = songSelect.value;
}}

songSelect.addEventListener('change', (e) => {{
    audioPlayer.src = e.target.value;
    audioPlayer.play();
}});

// MediaPipe Hands Detection
function onResults(results) {{
    canvasCtx.save();
    canvasCtx.clearRect(0, 0, canvasElement.width, canvasElement.height);

    // Draw mirrored video frame
    canvasCtx.translate(canvasElement.width, 0);
    canvasCtx.scale(-1, 1);
    canvasCtx.drawImage(results.image, 0, 0, canvasElement.width, canvasElement.height);

    if (results.multiHandLandmarks && results.multiHandLandmarks.length > 0) {{
        const landmarks = results.multiHandLandmarks[0];

        // Thumb Tip (4) and Index Tip (8)
        const thumb = landmarks[4];
        const index = landmarks[8];

        const tx = thumb.x * canvasElement.width;
        const ty = thumb.y * canvasElement.height;
        const ix = index.x * canvasElement.width;
        const iy = index.y * canvasElement.height;

        // Draw fingertip markers
        canvasCtx.fillStyle = '#FF007F';
        canvasCtx.beginPath();
        canvasCtx.arc(tx, ty, 8, 0, 2 * Math.PI);
        canvasCtx.fill();

        canvasCtx.beginPath();
        canvasCtx.arc(ix, iy, 8, 0, 2 * Math.PI);
        canvasCtx.fill();

        // Draw line between thumb and index
        canvasCtx.strokeStyle = '#00FF00';
        canvasCtx.lineWidth = 3;
        canvasCtx.beginPath();
        canvasCtx.moveTo(tx, ty);
        canvasCtx.lineTo(ix, iy);
        canvasCtx.stroke();

        // Calculate Euclidean distance
        const dist = Math.hypot(ix - tx, iy - ty);

        // Map distance (approx 20px - 160px) to volume (0.0 to 1.0)
        let vol = (dist - 20) / 140;
        vol = Math.max(0, Math.min(1, vol));

        audioPlayer.volume = vol;
        const pct = Math.round(vol * 100);
        volDisplay.innerText = "Volume: " + pct + "%";

        // Draw volume level indicator on canvas
        canvasCtx.restore();
        canvasCtx.fillStyle = '#222';
        canvasCtx.fillRect(10, 50, 15, 200);
        canvasCtx.fillStyle = '#00E676';
        canvasCtx.fillRect(10, 50 + (200 * (1 - vol)), 15, 200 * vol);
    }} else {{
        canvasCtx.restore();
    }}
}}

const hands = new Hands({{
    locateFile: (file) => `https://cdn.jsdelivr.net/npm/@mediapipe/hands/${{file}}`
}});

hands.setOptions({{
    maxNumHands: 1,
    modelComplexity: 1,
    minDetectionConfidence: 0.7,
    minTrackingConfidence: 0.7
}});

hands.onResults(onResults);

const camera = new Camera(videoElement, {{
    onFrame: async () => {{
        await hands.send({{image: videoElement}});
    }},
    width: 480,
    height: 360
}});

camera.start();
</script>
"""

display(HTML(html_code))
```
